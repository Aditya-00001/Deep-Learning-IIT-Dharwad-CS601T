```python
import os
import copy
import json
import numpy as np
import random
import torch
from torch import nn
from torch.utils.data import DataLoader
from torch.optim import SGD, Adam, Adagrad, RMSprop
from torchvision import datasets, transforms

if torch.cuda.is_available():
    device = torch.device("cuda")
    print("Using GPU:", torch.cuda.get_device_name(0))
else:
    device = torch.device("cpu")
    print("Using CPU")


LEARNING_RATE = 0.001
MOMENTUM = 0.9
epochs = 10000
THRESHOLD = 1e-4

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)


ACTIVATION = "tanh"

criterion = nn.CrossEntropyLoss()

ARCHITECTURES = {
    4: {
        "config_1": [128, 64, 32, 16],
        "config_2": [256, 128, 64, 32],
        "config_3": [512, 256, 128, 64]
    }
}
```

```python
data_transforms = transforms.Compose([
    transforms.Grayscale(num_output_channels=1),
    transforms.ToTensor()
])

train_data = datasets.ImageFolder(
    'Group_16/train',
    transform=data_transforms
)

val_data = datasets.ImageFolder(
    'Group_16/val',
    transform=data_transforms
)

test_data = datasets.ImageFolder(
    'Group_16/test',
    transform=data_transforms
)

print(train_data.class_to_idx)
print(val_data.class_to_idx)
print(test_data.class_to_idx)

assert train_data.class_to_idx == val_data.class_to_idx
assert train_data.class_to_idx == test_data.class_to_idx

print("Class mapping verified:", train_data.class_to_idx)

print("Training samples:", len(train_data))
print("Validation samples:", len(val_data))
print("Test samples:", len(test_data))

image, label = train_data[0]

print("Image shape:", image.shape)
```

# Optimizers and corresponding configs that will be used for each model

```python
def sgd(model):
  return SGD(model.parameters(),
             lr=LEARNING_RATE)

def sgd_general_delta(model):
  return SGD(model.parameters(),
             lr=LEARNING_RATE,
             momentum=MOMENTUM)

def sgd_nag(model):
  return SGD(model.parameters(),
             lr=LEARNING_RATE,
             momentum=MOMENTUM,
             nesterov=True)

def adagrad(model):
  return Adagrad(model.parameters(),
                 lr=LEARNING_RATE,
                 eps=1e-08
                 )

def rmsprop(model):
  return RMSprop(model.parameters(),
                lr=LEARNING_RATE,
                eps=1e-08,
                alpha=0.99,
                )

def adam(model):
  return Adam(model.parameters(),
              lr=LEARNING_RATE,
              betas=(0.9, 0.999),
              eps=1e-08
              )

optimizers={
    'sgd': sgd,
    'vanillaGD':sgd,
    'sgd_general_delta': sgd_general_delta,
    'sgd_nag': sgd_nag,
    'adagrad': adagrad,
    'rmsprop': rmsprop,
    'adam':adam
}

optimizer_config={
    'sgd':{
        'batch_size':1
    },
    'vanillaGD':{
        'batch_size':len(train_data)
    },
    'sgd_general_delta':{
        'batch_size':1
    },
    'sgd_nag':{
        'batch_size':1
    },
    'adagrad':{
        'batch_size':len(train_data)
    },
    'rmsprop':{
        'batch_size':len(train_data)
    },
    'adam':{
        'batch_size':1
    }
}
```

```python
def build_model(hidden_nodes):

    layers = [nn.Flatten()]

    for i in range(len(hidden_nodes)):

        if i == 0:
            layers.append(
                nn.Linear(784, hidden_nodes[i])
            )
        else:
            layers.append(
                nn.Linear(
                    hidden_nodes[i - 1],
                    hidden_nodes[i]
                )
            )

        layers.append(nn.Tanh())

    layers.append(
        nn.Linear(hidden_nodes[-1], 5)
    )

    return nn.Sequential(*layers)
```

```python
os.makedirs("initial", exist_ok=True)

for architecture in ARCHITECTURES:

    for config_name, hidden_nodes in ARCHITECTURES[architecture].items():

        output_path = (
            f"initial/model_{architecture}_{config_name}.pth"
        )

        if os.path.exists(output_path):

            print(f"Already exists: {output_path}")

        else:

            model = build_model(hidden_nodes)

            torch.save(
                model.state_dict(),
                output_path
            )

            print(f"Created: {output_path}")
```

```python
def calculate_average_error(model, data):

    data_loader = DataLoader(
        data,
        batch_size=128,
        shuffle=False,
        pin_memory=(device.type == "cuda")
    )

    model.eval()

    total_loss = 0.0
    total_samples = 0

    with torch.no_grad():

        for x, y in data_loader:

            x = x.to(device, non_blocking=True)
            y = y.to(device, non_blocking=True)

            output = model(x)

            loss = criterion(output, y)

            batch_size = y.size(0)

            total_loss += loss.item() * batch_size
            total_samples += batch_size

    return total_loss / total_samples
```

```python
def classification_accuracy(model, data):

    data_loader = DataLoader(
        data,
        batch_size=128,
        shuffle=False,
        pin_memory=(device.type == "cuda")
    )

    model.eval()

    correct = 0
    total = 0

    with torch.no_grad():

        for x, y in data_loader:

            x = x.to(
                device,
                non_blocking=True
            )

            y = y.to(
                device,
                non_blocking=True
            )

            output = model(x)

            prediction = torch.argmax(
                output,
                dim=1
            )

            correct += (
                prediction == y
            ).sum().item()

            total += y.size(0)

    return 100 * correct / total
```

```python
def run_optimizer(optimizer_name):

    architecture_id = list(ARCHITECTURES.keys())[0]

    # =========================================================
    # DIRECTORIES
    # =========================================================

    os.makedirs("models", exist_ok=True)
    os.makedirs("json", exist_ok=True)


    json_filename = (
        f"json/model_{architecture_id}_{optimizer_name}.json"
    )

    # =========================================================
    # LOAD EXISTING JSON IF IT EXISTS
    # This allows the notebook to resume after interruption.
    # =========================================================

    if os.path.exists(json_filename):

        print(f"Existing result file found: {json_filename}")
        print("Completed configurations will be skipped.")

        with open(json_filename, "r") as f:
            results = json.load(f)

    else:

        results = {
            "architecture": architecture_id,
            "activation": "tanh",
            "optimizer": optimizer_name,
            "learning_rate": LEARNING_RATE,
            "momentum": MOMENTUM,
            "threshold": THRESHOLD,
            "epochs_limit": epochs,
            "batch_size": optimizer_config[optimizer_name]["batch_size"],
            "configurations": {}
        }

    # =========================================================
    # START
    # =========================================================

    print("=" * 90)
    print(
        f"OPTIMIZER: {optimizer_name.upper()} | "
        f"ARCHITECTURE: {architecture_id} | "
        f"ACTIVATION: TANH"
    )
    print("=" * 90)

    # =========================================================
    # RUN ALL 3 CONFIGURATIONS
    # =========================================================

    for config_name, hidden_nodes in ARCHITECTURES[architecture_id].items():

        # -----------------------------------------------------
        # RESUME CHECK
        # -----------------------------------------------------

        if config_name in results["configurations"]:

            print("\n" + "-" * 90)
            print(
                f"SKIPPING {config_name} "
                f"(already completed)"
            )
            print("-" * 90)

            existing_result = results["configurations"][config_name]

            print(
                f"Saved Epochs: "
                f"{existing_result['epochs']}"
            )

            print(
                f"Saved Validation Accuracy: "
                f"{existing_result['final_val_accuracy']:.2f}%"
            )

            continue

        # -----------------------------------------------------
        # CONFIGURATION INFORMATION
        # -----------------------------------------------------

        print("\n" + "-" * 90)

        print(
            f"Architecture: {architecture_id} | "
            f"Configuration: {config_name} | "
            f"Neurons: {hidden_nodes} | "
            f"Optimizer: {optimizer_name}"
        )

        print("-" * 90)

        # =====================================================
        # BUILD MODEL
        # =====================================================

        model = build_model(hidden_nodes)

        # =====================================================
        # LOAD SAME INITIAL WEIGHTS
        # =====================================================

        initial_model_path = (
            f"initial/model_{architecture_id}_{config_name}.pth"
        )

        if not os.path.exists(initial_model_path):

            raise FileNotFoundError(
                f"Initial model not found:\n"
                f"{initial_model_path}"
            )

        model.load_state_dict(
            torch.load(
                initial_model_path,
                map_location=device
            )
        )

        model = model.to(device)

        # =====================================================
        # INITIAL ERROR
        # =====================================================

        initial_error = calculate_average_error(
            model,
            train_data
        )

        print(
            f"Initial Error: "
            f"{initial_error:.8f}"
        )

        # =====================================================
        # CREATE OPTIMIZER
        # =====================================================

        optimizer = optimizers[optimizer_name](model)

        # =====================================================
        # DATALOADER
        # =====================================================

        generator = torch.Generator()
        generator.manual_seed(SEED)

        train_loader = DataLoader(
            train_data,
            batch_size=optimizer_config[
                optimizer_name
            ]["batch_size"],
            shuffle=True,
            generator=generator,
            pin_memory=(device.type == "cuda")
        )

        # =====================================================
        # ERROR HISTORY
        #
        # errors[0] = error before training
        # errors[1] = error after epoch 1
        # etc.
        # =====================================================

        errors = [initial_error]

        # =====================================================
        # TRAINING
        # =====================================================

        for epoch in range(epochs):

            model.train()

            training_loss = 0.0
            total_samples = 0

            for x, y in train_loader:

                x = x.to(
                    device,
                    non_blocking=True
                )

                y = y.to(
                    device,
                    non_blocking=True
                )

                optimizer.zero_grad()

                output = model(x)

                loss = criterion(
                    output,
                    y
                )

                loss.backward()

                optimizer.step()

                current_batch_size = y.size(0)

                training_loss += (
                    loss.item() *
                    current_batch_size
                )

                total_samples += current_batch_size

            # -------------------------------------------------
            # AVERAGE TRAINING ERROR
            # -------------------------------------------------

            average_error = (
                training_loss /
                total_samples
            )

            errors.append(average_error)

            # -------------------------------------------------
            # SUCCESSIVE EPOCH ERROR DIFFERENCE
            # -------------------------------------------------

            error_difference = abs(
                errors[-1] -
                errors[-2]
            )

            # -------------------------------------------------
            # PROGRESS
            # -------------------------------------------------

            if (
                epoch == 0
                or (epoch + 1) % 10 == 0
            ):

                print(
                    f"Epoch {epoch + 1}: "
                    f"Error = {average_error:.8f} | "
                    f"Difference = "
                    f"{error_difference:.8f}"
                )

            # -------------------------------------------------
            # CONVERGENCE
            #
            # Assignment:
            # |E(t) - E(t-1)| < 1e-4
            # -------------------------------------------------

            if epoch>=2 and error_difference < THRESHOLD:

                print(
                    f"Converged at epoch "
                    f"{epoch + 1}"
                )

                break

        # =====================================================
        # FINAL METRICS
        # =====================================================

        final_error = errors[-1]

        converged_epochs = len(errors) - 1

        final_train_accuracy = (
            classification_accuracy(
                model,
                train_data
            )
        )

        final_val_accuracy = (
            classification_accuracy(
                model,
                val_data
            )
        )

        # =====================================================
        # SAVE TRAINED MODEL
        # =====================================================

        model_path = (
            f"models/"
            f"model_{architecture_id}_"
            f"{optimizer_name}_"
            f"{config_name}.pth"
        )

        torch.save(
            model.state_dict(),
            model_path
        )

        print(
            f"Saved trained model:\n"
            f"{model_path}"
        )

        # =====================================================
        # SAVE CONFIGURATION RESULT
        # =====================================================

        results["configurations"][config_name] = {

            "hidden_nodes": hidden_nodes,

            "initial_error": initial_error,

            "errors": errors,

            "epochs": converged_epochs,

            "final_error": final_error,

            "final_train_accuracy":
                final_train_accuracy,

            "final_val_accuracy":
                final_val_accuracy,

            "model_path": model_path
        }

        # =====================================================
        # SAVE JSON IMMEDIATELY
        #
        # IMPORTANT:
        # If Kaggle disconnects here, previous configurations
        # are already safely stored.
        # =====================================================

        with open(
            json_filename,
            "w"
        ) as f:

            json.dump(
                results,
                f,
                indent=4
            )

        print(
            f"Saved progress to:\n"
            f"{json_filename}"
        )

        print(
            f"Final Error: "
            f"{final_error:.8f}"
        )

        print(
            f"Final Train Accuracy: "
            f"{final_train_accuracy:.2f}%"
        )

        print(
            f"Final Validation Accuracy: "
            f"{final_val_accuracy:.2f}%"
        )

        print(
            f"Converged Epochs: "
            f"{converged_epochs}"
        )

        # =====================================================
        # FREE MEMORY BEFORE NEXT CONFIGURATION
        # =====================================================

        del model
        del optimizer
        del train_loader

        if device.type == "cuda":
            torch.cuda.empty_cache()

    # =========================================================
    # FINAL SUMMARY
    # =========================================================

    print("\n" + "=" * 90)

    print(
        f"COMPLETED OPTIMIZER: "
        f"{optimizer_name.upper()}"
    )

    print(
        f"Result file: "
        f"{json_filename}"
    )

    print("=" * 90)

    return results
```

```python
import matplotlib.pyplot as plt
import os
import json


# ============================================================
# LOAD ALL OPTIMIZER RESULTS
# ============================================================

def load_optimizer_results():

    architecture_id = list(ARCHITECTURES.keys())[0]

    all_results = {}

    for optimizer_name in optimizer_config:

        json_filename = (
            f"model_{architecture_id}_"
            f"{optimizer_name}.json"
        )

        if os.path.exists(json_filename):

            with open(
                json_filename,
                "r"
            ) as f:

                all_results[optimizer_name] = json.load(f)

        else:

            print(
                f"WARNING: "
                f"{json_filename} not found"
            )

    return all_results


# ============================================================
# PLOT TRAINING ERROR FOR EACH CONFIGURATION
# ============================================================

def plot_training_errors():

    architecture_id = list(ARCHITECTURES.keys())[0]

    all_results = load_optimizer_results()

    for config_name, hidden_nodes in ARCHITECTURES[
        architecture_id
    ].items():

        plt.figure(figsize=(10, 6))

        for optimizer_name in optimizer_config:

            if optimizer_name not in all_results:
                continue

            optimizer_results = all_results[
                optimizer_name
            ]

            if config_name not in optimizer_results[
                "configurations"
            ]:
                continue

            errors = optimizer_results[
                "configurations"
            ][
                config_name
            ][
                "errors"
            ]

            epochs_axis = range(
                len(errors)
            )

            plt.plot(
                epochs_axis,
                errors,
                label=optimizer_name
            )

        plt.xlabel("Epochs")

        plt.ylabel(
            "Average Training Error"
        )

        plt.title(
            f"Architecture {architecture_id} | "
            f"{config_name} | "
            f"Neurons: {hidden_nodes} | "
            f"Tanh"
        )

        plt.legend()

        plt.grid(True)

        plt.tight_layout()

        plt.show()


# ============================================================
# RUN PLOTS
# ============================================================

plot_training_errors()
```
